### 3. Countplots (Categorical & Binned Analysis)

```python
# Countplot 1: Distribution of observations across operating years
plt.figure(figsize=(7, 4))
sns.countplot(x='year', data=df, palette='Blues_d')
plt.title("Count of Hourly Records per Year (2011 - 2015)")
plt.xlabel("Operating Year")
plt.ylabel("Record Count")
plt.show()
```

```python
# Countplot 2: Categorizing NOx emission severity levels
df['NOX_Level'] = pd.cut(df['NOX'], bins=[0, 55, 75, 150], labels=['Low (<55)', 'Medium (55-75)', 'High (>75)'])

plt.figure(figsize=(7, 4))
sns.countplot(x='NOX_Level', data=df, palette='viridis')
plt.title("Countplot of NOx Emission Levels")
plt.xlabel("Emission Level (mg/m³)")
plt.ylabel("Count")
plt.show()
```

### 4. Histograms (Univariate Analysis)

```python
# Univariate Analysis
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

sns.histplot(df['AT'], kde=True, ax=axes[0, 0], color='blue')
axes[0, 0].set_title("Ambient Temperature (AT)")

sns.histplot(df['TIT'], kde=True, ax=axes[0, 1], color='green')
axes[0, 1].set_title("Turbine Inlet Temperature (TIT)")

sns.histplot(df['CDP'], kde=True, ax=axes[0, 2], color='red')
axes[0, 2].set_title("Compressor Discharge Pressure (CDP)")

sns.histplot(df['AFDP'], kde=True, ax=axes[1, 0], color='purple')
axes[1, 0].set_title("Air Filter Diff Pressure (AFDP)")

sns.histplot(df['TEY'], kde=True, ax=axes[1, 1], color='orange')
axes[1, 1].set_title("Turbine Energy Yield (TEY)")

sns.histplot(df['NOX'], kde=True, ax=axes[1, 2], color='black')
axes[1, 2].set_title("Target: Nitrogen Oxides (NOX)")

plt.suptitle("Feature Histograms with KDE", fontsize=14)
plt.tight_layout()
plt.show()
```

### 5. Outlier Detection with Boxplots (Individual Scales)

```python
# Using individual subplots so each feature auto-scales to its own range
fig, axes = plt.subplots(2, 5, figsize=(16, 7))
axes = axes.flatten()

features = ['AT', 'AP', 'AH', 'AFDP', 'GTEP', 'TIT', 'TAT', 'TEY', 'CDP', 'NOX']

for i, col in enumerate(features):
    sns.boxplot(y=df[col], ax=axes[i], color='skyblue')
    axes[i].set_title(col, fontweight='bold')

plt.suptitle('Boxplots of Features (Individual Scales)', fontsize=14)
plt.tight_layout()
plt.show()
```

I am not removing outlier rows because they represent real peak turbine operating conditions (cold weather starts, peak load). Tree ensembles (Random Forest, XGBoost) and SVR handle them naturally without deleting data.

### 6. Correlation Heatmap

```python
# Pairplots are not required here (redundant for 10 continuous features with 36k rows)
# The correlation heatmap and direct scatter plots give clear relationships
plt.figure(figsize=(10, 8))
corr = df[features].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

### 7. Bivariate Analysis: Top Features vs NOX

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.scatterplot(data=df.sample(1000, random_state=42), x='AT', y='NOX', ax=axes[0], color='blue')
axes[0].set_title("Ambient Temp vs NOX (r = -0.56)")

sns.scatterplot(data=df.sample(1000, random_state=42), x='TIT', y='NOX', ax=axes[1], color='green')
axes[1].set_title("Turbine Inlet Temp vs NOX (r = -0.21)")

sns.scatterplot(data=df.sample(1000, random_state=42), x='CDP', y='NOX', ax=axes[2], color='red')
axes[2].set_title("Compressor Pressure vs NOX (r = -0.17)")

plt.tight_layout()
plt.show()
```

### 8. Data Preprocessing & Saving Clean Dataset

```python
# Explicitly dropping year, CO (another pollutant output), and temporary NOX_Level column
df_processed = df.drop(columns=['year', 'CO', 'NOX_Level'], errors='ignore')

# Saving clean processed dataset for model training
df_processed.to_csv("dataset/gas_turbine_processed.csv", index=False)

print("Clean processed dataset saved to dataset/gas_turbine_processed.csv")
print("Remaining Columns (9 features + 1 target):", list(df_processed.columns))
print("Final Shape:", df_processed.shape)
df_processed.head()
```
